# CivicPulse — BBMP Grievance Intelligence

This notebook captures the end-to-end civic analytics workflow for BBMP complaint data.


```python
from pathlib import Path
import pandas as pd
base = Path('.')
raw_dir = base / 'data' / 'raw'
frames = [pd.read_csv(f) for f in sorted(raw_dir.glob('*.csv'))]
df = pd.concat(frames, ignore_index=True)
df['Grievance Date'] = pd.to_datetime(df['Grievance Date'], errors='coerce')
df = df.dropna(subset=['Grievance Date']).copy()
print('Rows:', len(df))
print('Years:', sorted(df['Grievance Date'].dt.year.unique().tolist()))
print('Top categories:', df['Category'].value_counts().head().to_dict())
```

## Key project insight

Electrical complaints and municipal infrastructure issues dominate the complaint profile across 2020–2024.
